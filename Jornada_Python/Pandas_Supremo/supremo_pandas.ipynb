```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("vendas_tech.csv", encoding="utf-8", low_memory=False)
display(df_vendas)
df_gerentes = pd.read_excel("gerentes_lojas.xlsx")
display(df_gerentes)
```

```python
#inspeção geral
display(df_vendas)
#inspeção () 5 primeiras linhas do arquivo ou (10)
display(df_vendas.head(10))
#inspeção final do arquivo () 5 primeiras linhas do arquivo ou (10) 
display(df_vendas.tail(10))
#inspeção dados aleatórios
display(df_vendas.sample(10))
#saber a quantidade de linhas do arquivo
display(df_vendas.shape)
#ver todas as colunas do arquivo
display(df_vendas.columns) 
#ver todas as colunas do arquivo como lista
display(list(df_vendas.columns))
#info e describe
display(df_vendas.info())
display(df_vendas.describe())
```

```python
#tratamento de dados
#display(df_vendas["Loja"])
#display(df_vendas[["Loja","Cliente"]])
#colunas
df_analise = df_vendas.drop(columns=["Data_Base"])
#joga fora da tabela os nulos
#df_analise = df_analise.dropna()
#df_analise["Loja"] = df_analise["Loja"].fillna("Online")
#tipos de dados
df_analise["Data"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d", errors="coerce")
#padronização
#tirar espaço início e final
#df_analise["Loja"] = df_analise["Loja"].str.strip()
#primeira letra maiúscula
#df_analise["Loja"] = df_analise["Loja"].str.title()
#encadeamento
df_analise["Loja"] = df_analise["Loja"].fillna("Online").str.strip().str.title()
#registro duplicado
df_analise = df_analise.drop_duplicates(subset=["ID_Pedido", "Loja"],keep="first")
display(df_analise)
display(df_analise.info())
```

```python
#criar nova coluna
#faturamento
df_analise["Faturamento"] = df_analise["Qtd"] * df_analise["Preco_Unitario"] 
#forma de venda
df_analise["Forma_de_Vendas"] = np.where(df_analise["Loja"]=="Online", "Online", "Presencial")
#região
display(df_analise["Loja"].unique()) 
#dicionário de de para  
dic_regioes = {
    'São Paulo': "Sudeste",
    'Belo Horizonte': "Sudeste", 
    'Online': "Online",
    'Rio De Janeiro': "Sudeste", 
    'Salvador': "Nordeste", 
    'Recife': "Nordeste",  
    'Curitiba': "Sul",  
    'Porto Alegre': "Sul"
}

df_analise["Região"] = df_analise["Loja"].map(dic_regioes)
                                
display(df_analise)
#mostra quantos valores vazios/nulos existem em cada coluna do DataFrame df_analise
display(df_analise.isna().sum())
```

```python
#analisar - filtrar
#ordernar a base pelo campo Data
#Faturamente ordenado decrescente
df_analise = df_analise.sort_values(
    by=["Data", "Faturamento"],
    ascending=[True, False]
)
#vai recriar a coluna de index ordenado menor para maior
#exclui o nome index da coluna
df_analise = df_analise.reset_index(drop=True)
#.loc: id pedido - localiza por nome da coluna e linha
id_pedido = 4
loja = df_analise.loc[3, "Loja"]
produto = df_analise.loc[3, "Produto"]
cliente = df_analise.loc[3, "Cliente"]
print(loja, produto, cliente)
#.iloc: id pedido - localiza por posição
id_pedido = 4
loja = df_analise.iloc[3, 2]
produto = df_analise.iloc[3, 3]
cliente = df_analise.iloc[3, 6]
print(loja, produto, cliente)
display(df_analise)

#exportar pedaços da base
#df_vendas_sp = df_analise["Loja"]
```
